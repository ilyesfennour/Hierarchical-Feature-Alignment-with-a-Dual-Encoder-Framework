```python
from main import run_training

EPOCHS = 50
BATCH_SIZE = 128
LEARNING_RATE = 0.0002
WORKERS = 8
NUM_LAYERS = 4

AUTOENCODER='DFA'            # DFA, LAE, DLAE
DATASET = 'MNIST'            # MedMNIST, ImageNet, MNIST, ImageNet, CelebA, SVHN, DFB-HR
ROOT_DIR = None
DEGRADATION = 'denoising'    # denoising, super_resolution, missing_data, None(for clustering and classification)
MAX_SAMPLES_PER_CLASS = None
PRETRAINED_PATH = None

SIGMA = 75/255.0
SCALE_FACTOR = 4

LOSSES_CONFIG = {
    'l1': {'lambda': 1.0},
    'l2': {'lambda': 0.5},
    'perceptual': {'lambda': 0.1},
    'tap': {'lambda': 0.05},
    'latent': {'lambda': 0.1},
    'ssim': {'lambda': 0.5},
    'edge': {'lambda': 0.1},
    'ce': {'lambda': 1.0},
    'center': {'lambda': 0.01},
    'triplet': {'lambda': 0.1}
}

CHECKPOINT_DIR = None
ADDITIONAL_EPOCHS = 0

# Start training
history, exp_dir = run_training(
    autoencoder=AUTOENCODER,
    losses_config=LOSSES_CONFIG,
    dataset=DATASET,
    root_dir=ROOT_DIR,
    degradation=DEGRADATION,
    max_samples_per_class=MAX_SAMPLES_PER_CLASS,
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    learning_rate=LEARNING_RATE,
    workers=WORKERS,
    num_layers=NUM_LAYERS,
    pretrained_path=PRETRAINED_PATH,
    checkpoint_dir = CHECKPOINT_DIR,
    additional_epochs = ADDITIONAL_EPOCHS,
    sigma=SIGMA,
    scale_factor=SCALE_FACTOR
)

# Print summary
if history and exp_dir:
    print(f"\n✅ Training completed successfully!")
    print(f"📁 Results saved to: {exp_dir}")
```
