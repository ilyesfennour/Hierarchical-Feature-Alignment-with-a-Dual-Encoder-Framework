```python
import sys
import os

sys.path.insert(0, os.path.abspath(".."))

from main import run_training

# Training parameters
EPOCHS = 50
BATCH_SIZE = 128
LEARNING_RATE = 2e-4
WORKERS = 8

# Model architecture
AUTOENCODER = 'DFA'       # DFA, LAE, DLAE
NUM_LAYERS = 4
TAP_LAYERS = 3
LATENT_DIM = 128

# Dataset configuration
DATASET = 'MNIST'         # MNIST, MedMNIST, CIFAR-10, SVHN, CelebA, ImageNet, DFB-HR
ROOT_DIR = None           # Custom dataset directory (if required)
MAX_SAMPLES_PER_CLASS = None

# Task configuration
DEGRADATION = 'denoising' # denoising, super_resolution, missing_data, None(for classification and clustering)
SIGMA = 75 / 255.0        # Gaussian noise standard deviation
SCALE_FACTOR = 4          # Super-resolution scaling factor

# Pretrained model
PRETRAINED_PATH = None

# Checkpoint and training continuation
CHECKPOINT_DIR = None
ADDITIONAL_EPOCHS = 0

# Loss Configuration
LOSSES_CONFIG = {
    'l1':        {'lambda': 1.0},
    'l2':        {'lambda': 0.5},
    'perceptual':{'lambda': 0.1},
    'tap':       {'lambda': 0.05},
    # 'latent':    {'lambda': 0.1},
    # 'ssim':      {'lambda': 0.5},
    # 'edge':      {'lambda': 0.1},
    # 'ce':        {'lambda': 1.0},
    # 'center':    {'lambda': 0.01},
    # 'triplet':   {'lambda': 0.1}
}

# Start training
history, exp_dir = run_training(
    autoencoder=AUTOENCODER,
    losses_config=LOSSES_CONFIG,
    dataset=DATASET,
    root_dir=ROOT_DIR,
    degradation=DEGRADATION,
    max_samples_per_class=MAX_SAMPLES_PER_CLASS,
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    learning_rate=LEARNING_RATE,
    workers=WORKERS,
    num_layers=NUM_LAYERS,
    tap_layers=TAP_LAYERS,
    latent_dim=LATENT_DIM,
    pretrained_path=PRETRAINED_PATH,
    checkpoint_dir=CHECKPOINT_DIR,
    additional_epochs=ADDITIONAL_EPOCHS,
    sigma=SIGMA,
    scale_factor=SCALE_FACTOR
)

# Print summary
if history and exp_dir:
    print(f"\n✅ Training completed successfully!")
    print(f"📁 Results saved to: {exp_dir}")
```
